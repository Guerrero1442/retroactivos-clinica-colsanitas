```python
import pandas as pd
import pyarrow as pa
from pathlib import Path
```

```python
string_pa = pd.ArrowDtype(pa.string())
```

```python
upc_columns = {
    "prestador": string_pa,
    "depto_prestador": string_pa,
    "mun_prestador": string_pa,
    "no_id": string_pa,
    "tipo_id": string_pa,
    "no_factura": string_pa,
}
```

```python
upc_data_path = Path(r"..\outputs\upc_data.csv")
resultados_path = Path(r"..\outputs\resultados.csv")
```

```python
df_upc = pd.read_csv(upc_data_path, sep="|", dtype=upc_columns, dtype_backend="pyarrow")
df_upc.head()
```

```python
columns_resultado = {
    "num factura": string_pa,
    "tip doc": string_pa,
    "num doc": string_pa,
    "Codigo Cups": string_pa,
}
```

```python
df_resultado = pd.read_csv(
    resultados_path, sep="|", dtype=columns_resultado, dtype_backend="pyarrow"
)
df_resultado
```

```python
df_upc.loc[
    df_upc.duplicated(
        subset=["no_factura", "tipo_id", "no_id", "cups", "f_prestacion"], keep=False
    )
].sort_values(
    by=["no_factura", "tipo_id", "no_id", "cups", "f_prestacion"], ascending=[True] * 5
)[["no_factura", "tipo_id", "no_id", "cups", "desc_cups", "f_prestacion", "vr_pagado"]]
```

```python
df_tarifas_2025 = pd.read_excel(
    r"..\inputs\30348_ANEXO_800149384_CLINICA KERALTY IBAGUE (2025).xlsx",
    sheet_name="ANEXO",
    skiprows=18,
)

df_tarifas_2025.head()
```

```python
df_merge_upc = df_upc.merge(
    df_tarifas_2025.dropna(subset=["COD EPS"]),
    how="left",
    left_on="cups",
    right_on="COD EPS",
    suffixes=("_upc", "_tarifa2025"),
)

df_merge_upc.head()
```

```python
df_merge_upc.loc[
    df_merge_upc.duplicated(
        subset=["no_factura", "tipo_id", "no_id", "cups", "f_prestacion"], keep=False
    )
].sort_values(
    by=["no_factura", "tipo_id", "no_id", "cups", "f_prestacion"], ascending=[True] * 5
)[
    [
        "no_factura",
        "tipo_id",
        "no_id",
        "cups",
        "desc_cups",
        "f_prestacion",
        "vr_pagado",
    ]
]
```

```python
agrupacion_costo_upc = df_upc.groupby(
    ["no_factura", "tipo_id", "no_id", "cups", "f_prestacion"], as_index=False
).agg({"vr_pagado": "sum"})

agrupacion_costo_upc
```

```python
df_resultado_merge_upc = df_resultado.merge(
    agrupacion_costo_upc,
    how="left",
    left_on=["num factura", "tip doc", "num doc", "Codigo Cups", "fec cargo"],
    right_on=["no_factura", "tipo_id", "no_id", "cups", "f_prestacion"],
    suffixes=("_resultado", "_upc"),
)

df_resultado_merge_upc.drop(
    columns=["no_factura", "tipo_id", "no_id", "cups", "f_prestacion"], inplace=True
)

df_resultado_merge_upc.head()
```

```python
df_tarifas_2024_27000 = pd.read_excel(
    r"..\inputs\27000_ANEXO_800149384_CLINICA KERALTY IBAGUE_(2024).xlsx",
    sheet_name="ANEXO",
    skiprows=17,
    dtype_backend="pyarrow",
)

df_tarifas_2024_27000.head()
```

```python
df_tarifas_2024_27467 = pd.read_excel(
    r"..\inputs\27467_ANEXO_800149384_CLINICA KERALTY IBAGUE_(2024).xlsx",
    sheet_name="ANEXO",
    skiprows=17,
    dtype_backend="pyarrow",
)

df_tarifas_2024_27467.head()
```

```python
df_tarifas_2024_27467.dropna(subset=["COD EPS"], inplace=True)
df_tarifas_2024_27000.dropna(subset=["COD EPS"], inplace=True)
```

```python
df_tarifas_2024 = pd.concat(
    [df_tarifas_2024_27000, df_tarifas_2024_27467], ignore_index=True
)

df_tarifas_2024.head()
```

```python
df_resultado_merge_tarifas_2024 = df_resultado_merge_upc.merge(
    df_tarifas_2024,
    how="left",
    left_on="Codigo Cups",
    right_on="COD EPS",
    suffixes=("_upc", "_tarifa2024"),
)
```

```python
df_resultado_merge_tarifas_2025 = df_resultado_merge_tarifas_2024.merge(
    df_tarifas_2025.dropna(subset=["COD EPS"]),
    how="left",
    left_on="Codigo Cups",
    right_on="COD EPS",
    suffixes=("_tarifa2024", "_tarifa2025"),
)
```

```python
cols_salida = [
    "num factura",
    "desc sucur",
    "tip doc",
    "num doc",
    "pri nom",
    "seg nom",
    "pri ape",
    "seg ape",
    "tip admision",
    "ano adm",
    "num adm",
    "desc entidad",
    "fec cargo",
    "ano cargo",
    "mes cargo",
    "Codigo Cups",
    "desc insumo servicio",
    "desc concepto",
    "valor unit",
    "cantidad",
    "val total cargo",
    "fecha factura",
    "Fecha Ingreso",
    "Fecha Egreso",
    "CODIGO BH",
    "TARIFA 2025",
    "TARIFA VRS CANTIDAD",
    "VALOR RETROACTIVO",
    "llave_retroactivo",
    "encontrado_upc",
    "vr_pagado",
    "TOTAL_tarifa2024",
    "TOTAL_tarifa2025",
]
```

```python
df_salida_resultado_ibague = df_resultado_merge_tarifas_2025[cols_salida].copy()
```

```python
df_salida_resultado_ibague
```

## Operaciones


```python
# vr_pagado upc / total_2024 = porcentaje_upc_costo_2024
# porcentaje_upc_costo_2024 * total_2025 = vr_pagado 2025
# vr_pagado 2025 - vr_pagado upc = retroactivo
```

```python
df_salida_resultado_ibague = df_salida_resultado_ibague.assign(
    porcentaje_upc_costo_2024=lambda df_: df_["vr_pagado"] / df_["TOTAL_tarifa2024"],
    vr_pagado_2025=lambda df_: df_["porcentaje_upc_costo_2024"]
    * df_["TOTAL_tarifa2025"],
    valor_retroactivo=lambda df_: df_["vr_pagado_2025"] - df_["vr_pagado"],
)
```

```python
# caracterizas 30mil registros que no cruzaron con upc
df_salida_resultado_ibague_no_upc = df_salida_resultado_ibague.loc[
    df_salida_resultado_ibague["encontrado_upc"] == "NO"
]
```

```python
# caracterizas 30mil registros que no cruzaron con upc
df_salida_resultado_ibague_si_upc = df_salida_resultado_ibague.loc[
    df_salida_resultado_ibague["encontrado_upc"] == "SI"
]
```

# HOJAS

1.NO ENCONTRADOS EN UPC
2.TD NO ENCONTRADOS UPC (CUPS, FACTURA, F_CARGO)
3.retroactivo ibague > retroactivo calculado
4.retroactivo ibague < retroactivo calculado
5.retroactivo ibague = retroactivo calculado
6.duplicados


```python
subset_duplicados = [
    "num factura",
    "desc sucur",
    "tip doc",
    "num doc",
    "pri nom",
    "seg nom",
    "pri ape",
    "seg ape",
    "tip admision",
    "ano adm",
    "num adm",
    "desc entidad",
    "fec cargo",
    "ano cargo",
    "mes cargo",
    "Codigo Cups",
    "desc insumo servicio",
    "desc concepto",
    "valor unit",
    "cantidad",
    "val total cargo",
    "fecha factura",
    "Fecha Ingreso",
    "Fecha Egreso",
    "CODIGO BH",
    "TARIFA 2025",
    "TARIFA VRS CANTIDAD",
    "VALOR RETROACTIVO",
]
```

```python
df_duplicados = df_salida_resultado_ibague.loc[
    df_salida_resultado_ibague.duplicated(subset=subset_duplicados, keep=False)
]
```

```python
df_salida_resultado_ibague["DUPLICADO"] = df_salida_resultado_ibague.duplicated(
    subset=subset_duplicados, keep=False
)

df_salida_resultado_ibague = df_salida_resultado_ibague.assign(
    DUPLICADO=lambda df_: df_["DUPLICADO"].map({True: "DUPLICADO", False: "UNICO"}),
    diferencia_retroactivo=lambda df_: df_["VALOR RETROACTIVO"]
    - df_["valor_retroactivo"],
).dropna(subset=["Codigo Cups"])
```

```python
# 3.retroactivo ibague > retroactivo calculado
# 4.retroactivo ibague < retroactivo calculado
# 5.retroactivo ibague = retroactivo calculado

df_retroactivo_ibague_gt_calculado = df_salida_resultado_ibague.loc[
    df_salida_resultado_ibague["VALOR RETROACTIVO"].gt(
        df_salida_resultado_ibague["valor_retroactivo"]
    )
]

df_retroactivo_ibague_lt = df_salida_resultado_ibague.loc[
    df_salida_resultado_ibague["VALOR RETROACTIVO"].lt(
        df_salida_resultado_ibague["valor_retroactivo"]
    )
]

df_retroactivo_ibague_eq = df_salida_resultado_ibague.loc[
    df_salida_resultado_ibague["VALOR RETROACTIVO"].eq(
        df_salida_resultado_ibague["valor_retroactivo"]
    )
]
```

```python
with pd.ExcelWriter(
    r"..\outputs\analisis_retroactivo_ibague.xlsx",
    engine="openpyxl",
) as writer:
    df_salida_resultado_ibague.to_excel(
        writer, sheet_name="resultado_completo", index=False
    )

    df_duplicados.to_excel(writer, sheet_name="duplicados", index=False)

    df_retroactivo_ibague_gt_calculado.to_excel(
        writer, sheet_name="retroactivo_gt_calculado", index=False
    )
    df_retroactivo_ibague_lt.to_excel(
        writer, sheet_name="retroactivo_lt_calculado", index=False
    )
    df_retroactivo_ibague_eq.to_excel(
        writer, sheet_name="retroactivo_eq_calculado", index=False
    )
```
