## Operaciones


```python
# vr_neto upc / total_2024 = porcentaje_upc_costo_2024
# porcentaje_upc_costo_2024 * total_2025 = vr_neto 2025
# vr_neto 2025 - vr_neto upc = retroactivo
```

```python
df_salida_resultado_ibague = df_salida_resultado_ibague.assign(
    porcentaje_upc_costo_2024=lambda df_: df_["vr_neto"] / df_["TOTAL_tarifa2024"],
    vr_neto_2025=lambda df_: df_["porcentaje_upc_costo_2024"] * df_["TOTAL_tarifa2025"],
    valor_retroactivo=lambda df_: df_["vr_neto_2025"] - df_["vr_neto"],
)
```
